```julia
using JLD2
using CairoMakie
```

```julia
filename = "qsd_J1.0_γ0.5_L64_single_20251103_22-37-56.jld2"
filepath = "../output/" * filename
data = load(filepath)
```

```julia
x_ax = [log(t) for t in data["times"]];
y_ax = data["S_ent"];
```

```julia
fig = Figure(size=(800, 600))
ax = Axis(fig[1,1], xlabel="log(t)", ylabel="S_ent")

scatter!(ax, x_ax, y_ax, markersize=5, color=:orange)
xlims!(ax, -7, 5)

fig
```
